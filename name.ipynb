```python
import pandas as pd
import numpy as np
df = pd.DataFrame(np.random.randint(1, 100, size=(5, 3)),
columns=['A', 'B', 'C'])
df['Product'] = df['A'] * df['B']
print(df)
```

```python
import pandas as pd
import numpy as np
df = pd.DataFrame(np.random.randint(1, 100, size=(4, 3)),
columns=['A', 'B', 'C'])
row_sum = df.sum(axis=1)
column_sum = df.sum(axis=0)
print("DataFrame:")
print(df)
print("\nRow-wise Sum:")
print(row_sum)
print("\nColumn-wise Sum:")
print(column_sum)
```

```python
import pandas as pd
import numpy as np
df = pd.DataFrame(np.random.randint(1, 100, size=(5, 3)),
columns=['A', 'B', 'C'])
df.iloc[1, 0] = np.nan
df.iloc[2, 1] = np.nan
df.iloc[3, 2] = np.nan
print("Original DataFrame:")
print(df)
df = df.fillna(df.mean())
print("\nDataFrame after filling NaN values:")
print(df)
```

```python
import pandas as pd
import numpy as np
df = pd.DataFrame(np.random.randint(1, 100, size=(6, 4)),
columns=['A', 'B', 'C', 'D'])
df.iloc[1, 0] = np.nan
df.iloc[3, 2] = np.nan
df.iloc[4, 1] = np.nan
print("Original DataFrame:")
print(df)
df = df.dropna()
print("\nDataFrame after dropping NaN rows:")
print(df)
```

```python
import pandas as pd
import numpy as np
df = pd.DataFrame({
'Category': np.random.choice(['A', 'B', 'C'], size=10),
'Value': np.random.randint(1, 100, size=10)
})
print("Original DataFrame:")
print(df)
result = df.groupby('Category')['Value'].agg(['sum', 'mean'])
print("\nData Aggregation:")
print(result)
```

```python
import pandas as pd
import numpy as np
df = pd.DataFrame({
'Product': np.random.choice(['Laptop', 'Mobile', 'TV'], size=10),
'Category': np.random.choice(['Electronics', 'Clothing', 'Furniture'], size=10),
'Sales': np.random.randint(100, 1000, size=10)
})
print("Original DataFrame:")
print(df)
result = df.groupby('Category')['Sales'].sum()
print("\nTotal Sales by Category:")
print(result)
```

```python
import pandas as pd
df1 = pd.DataFrame({
'ID': [1, 2, 3],
'Name': ['Ali', 'razeen', 'swasthik']
})
df2 = pd.DataFrame({
'ID': [1, 2, 3],
'Marks': [85, 90, 95]
})
print("First DataFrame:")
print(df1)
print("\nSecond DataFrame:")
print(df2)
result = pd.merge(df1, df2, on='ID')
print("\nMerged DataFrame:")
print(result)
```

```python
import pandas as pd
index = pd.MultiIndex.from_tuples([
('A', 1), ('A', 2), ('B', 1), ('B', 2)
], names=['Category', 'Number'])
df = pd.DataFrame({
'Value': [10, 20, 30, 40]
}, index=index)
print("Original DataFrame:")
print(df)
print("\nAccessing Category A:")
print(df.loc['A'])
print("\nSlicing DataFrame:")
print(df.loc['A':'B'])
```

```python
import pandas as pd
import numpy as np
index = pd.MultiIndex.from_tuples([
('A','X'), ('A','Y'),
('B','X'), ('B','Y'),
('C','X'), ('C','Y')
], names=['Category', 'SubCategory'])
df = pd.DataFrame({
'Value': np.random.randint(1, 100, size=6)
}, index=index)
print("Original DataFrame:")
print(df)
result = df.groupby(['Category', 'SubCategory'])['Value'].sum()
print("\nSum of Values:")
print(result)
```

```python
import pandas as pd
import numpy as np
df = pd.DataFrame(np.random.randint(1, 100, size=(5, 3)),
columns=['A', 'B', 'C'])
print("Original DataFrame:")
print(df)
result = df.apply(lambda x: x * 2)
print("\nDataFrame after doubling:")
print(result)
```

```python
import pandas as pd
import numpy as np
df = pd.DataFrame(np.random.randint(1, 100, size=(6, 3)),
columns=['A', 'B', 'C'])
print("Original DataFrame:")
print(df)
df['Sum'] = df.apply(lambda x: x['A'] + x['B'] + x['C'], axis=1)
print("\nDataFrame after adding Sum column:")
print(df)
```
